# Dataset Description 


The Center Sector Scheme 2021-22 dataset provides a detailed financial overview of various schemes under different ministries and departments of the Indian government. This dataset includes budget allocations, actual expenditures, and revised estimates for multiple years. The analysis of this dataset helps in understanding trends in public spending, identifying key areas of investment, and evaluating the effectiveness of budget allocation.



# Dataset Overview:

Category – Identifies demand categories (e.g., Demand No. 1, Demand No. 2).

Sl.No. – Serial number of the scheme (may contain missing values).

Ministry/Department – The name of the government ministry or department implementing the scheme.

Scheme – The name of the central sector scheme under consideration.

Actuals2019-2020 - Revenue – The revenue expenditure incurred for the scheme in 2019-2020.

Actuals2019-2020 - Capital – The capital expenditure incurred for the scheme in 2019-2020.

Actuals2019-2020 - Total – The total expenditure (revenue + capital) for 2019-2020.

Budget Estimates2020-2021 - Revenue – Estimated revenue expenditure for 2020-2021.

Budget Estimates2020-2021 - Capital – Estimated capital expenditure for 2020-2021.

Budget Estimates2020-2021 - Total – Total estimated expenditure for 2020-2021.

Revised Estimates2020-2021 - Revenue – Revised revenue expenditure for 2020-2021.

Revised Estimates2020-2021 - Capital – Revised capital expenditure for 2020-2021.

Revised Estimates2020-2021 - Total – Revised total expenditure for 2020-2021.

Budget Estimates2021-2022 - Revenue – Estimated revenue expenditure for 2021-2022.

Budget Estimates2021-2022 - Capital – Estimated capital expenditure for 2021-2022.

Budget Estimates2021-2022 - Total – Total estimated expenditure for 2021-2022.

```python
## df.fillna(df.mean(), inplace=True)  # Fill with mean
# Fill missing values only in numeric columns
df.fillna(df.select_dtypes(include=['number']).mean(), inplace=True)
df
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
df
```

```python
# Standardizing categorical text columns
df = df.apply(lambda x: x.str.strip().str.lower() if x.dtype == "object" else x)
df
```

```python
from sklearn.preprocessing import LabelEncoder

# Apply label encoding to categorical columns
label_encoder = LabelEncoder()
for col in df.select_dtypes(include=['object']).columns:
    df[col] = label_encoder.fit_transform(df[col])
df
```

```python
df = pd.get_dummies(df, drop_first=True)
df
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("CenterSectorScheme2021-22.csv")

# Convert relevant columns to numeric (handling errors)
df["Budget Estimates2021-2022 - Total"] = pd.to_numeric(df["Budget Estimates2021-2022 - Total"], errors='coerce')
df["Actuals2019-2020 - Total"] = pd.to_numeric(df["Actuals2019-2020 - Total"], errors='coerce')

# Calculate total sums for the two columns
budget_estimate = df["Budget Estimates2021-2022 - Total"].sum()
actual_expenditure = df["Actuals2019-2020 - Total"].sum()

# Data for pie chart
labels = ["Budget Estimate 2021-22", "Actual Expenditure 2019-20"]
values = [budget_estimate, actual_expenditure]

# Plot pie chart
plt.figure(figsize=(8, 6))
plt.pie(values, labels=labels, autopct='%1.1f%%', colors=["blue", "green"], startangle=140)
plt.title("Comparison of Budget Estimate vs. Actual Expenditure")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset
df = pd.read_csv("CenterSectorScheme2021-22.csv")

# Clean column names (remove extra spaces)
df.columns = df.columns.str.strip()

# Convert necessary columns to numeric (forcing errors='coerce' to convert non-numeric values to NaN)
df["Actuals2019-2020 - Total"] = pd.to_numeric(df["Actuals2019-2020 - Total"], errors='coerce')
df["Budget Estimates2021-2022 - Total"] = pd.to_numeric(df["Budget Estimates2021-2022 - Total"], errors='coerce')

# Drop NaN values that were caused by non-numeric data
df = df.dropna(subset=["Actuals2019-2020 - Total", "Budget Estimates2021-2022 - Total"])

# Regression Plot
plt.figure(figsize=(8, 6))
sns.regplot(data=df, x="Actuals2019-2020 - Total", y="Budget Estimates2021-2022 - Total", scatter_kws={'alpha':0.6}, line_kws={"color": "red"})

plt.title("Regression Plot of Actuals vs Budget Estimates")
plt.xlabel("Actuals 2019-2020 - Total")
plt.ylabel("Budget Estimates 2021-2022 - Total")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset
df = pd.read_csv("CenterSectorScheme2021-22.csv")

# Clean column names (remove extra spaces)
df.columns = df.columns.str.strip()

# Convert necessary columns to numeric
df["Actuals2019-2020 - Total"] = pd.to_numeric(df["Actuals2019-2020 - Total"], errors='coerce')
df["Budget Estimates2021-2022 - Total"] = pd.to_numeric(df["Budget Estimates2021-2022 - Total"], errors='coerce')

# Drop NaN values that were caused by non-numeric data
df = df.dropna(subset=["Actuals2019-2020 - Total", "Budget Estimates2021-2022 - Total"])

# Reshape data for box plot
df_melted = df.melt(value_vars=["Actuals2019-2020 - Total", "Budget Estimates2021-2022 - Total"], 
                     var_name="Year", value_name="Amount")

# Box Plot
plt.figure(figsize=(8, 6))
sns.boxplot(x="Year", y="Amount", data=df_melted, palette=["blue", "red"])

plt.title("Box Plot of Actuals vs Budget Estimates")
plt.xlabel("Year")
plt.ylabel("Amount (in Crores)")
plt.grid(True)
plt.show()
```

```python
# Line Chart - Trends Over Time
plt.figure(figsize=(10, 5))

# Sort data (Assuming schemes are in order)
df_sorted = df.sort_values(by="Actuals2019-2020 - Total")

plt.plot(df_sorted["Actuals2019-2020 - Total"], label="Actuals 2019-2020", marker="o", linestyle="-")
plt.plot(df_sorted["Budget Estimates2021-2022 - Total"], label="Budget Estimates 2021-2022", marker="s", linestyle="--")

plt.title("Line Chart: Actuals vs Budget Estimates")
plt.xlabel("Index")
plt.ylabel("Amount (in Crores)")
plt.legend()
plt.grid(True)
plt.show()
```
